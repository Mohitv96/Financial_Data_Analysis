```python
import numpy as np

# 1. Define constants
# Assuming 252 trading days in a year and a 0% risk-free rate for simplicity
risk_free_rate = 0 
trading_days = 252

# 2. Calculate Daily Statistics
stats = df_all.groupby('instrument')['percent_change'].agg(['mean', 'std'])

# 3. Calculate Daily Sharpe Ratio
stats['daily_sharpe'] = (stats['mean'] - risk_free_rate) / stats['std']

# 4. Annualize the Sharpe Ratio
stats['annualized_sharpe'] = stats['daily_sharpe'] * np.sqrt(trading_days)

# 5. Visualize the results
plt.figure(figsize=(10, 6))
sns.barplot(x=stats.index, y=stats['annualized_sharpe'], palette='viridis')

plt.title('Risk-Adjusted Return: Annualized Sharpe Ratio')
plt.ylabel('Sharpe Ratio')
plt.xlabel('Instrument')
plt.axhline(0, color='black', lw=1) # Baseline
plt.show()

print(stats[['annualized_sharpe']])
```

```python
# 1. Focus on one instrument for the backtest (NSE Index)
df_strategy = df_all[df_all['instrument'] == 'NSE Index'].copy()

# 2. Generate Signals
# Signal = 1 (Buy/Hold), Signal = 0 (Sell/Cash)
df_strategy['signal'] = np.where(df_strategy['ma_20'] > df_strategy['ma_50'], 1, 0)

# 3. Calculate Strategy Returns
# We shift the signal by 1 day because we can only act on the signal the NEXT day
df_strategy['strategy_return'] = df_strategy['signal'].shift(1) * df_strategy['percent_change']

# 4. Calculate Cumulative Returns
# This shows how $1 grows over time
df_strategy['cumulative_market_return'] = (1 + df_strategy['percent_change']/100).cumprod()
df_strategy['cumulative_strategy_return'] = (1 + df_strategy['strategy_return']/100).cumprod()

# 5. Visualize Strategy vs. Market
plt.figure(figsize=(12, 6))
plt.plot(df_strategy['date'], df_strategy['cumulative_market_return'], label='Buy & Hold (Market)', color='gray', alpha=0.5)
plt.plot(df_strategy['date'], df_strategy['cumulative_strategy_return'], label='MA Crossover Strategy', color='green', linewidth=2)

plt.title('Strategy Backtest: MA-20/50 Crossover vs. Buy & Hold')
plt.xlabel('Date')
plt.ylabel('Cumulative Growth of $1')
plt.legend()
plt.show()

# Final Performance Metric
final_market = df_strategy['cumulative_market_return'].iloc[-1]
final_strategy = df_strategy['cumulative_strategy_return'].iloc[-1]
print(f"Final Market Return: {round((final_market-1)*100, 2)}%")
print(f"Final Strategy Return: {round((final_strategy-1)*100, 2)}%")
```

```python
df_nsei=df_all[df_all['instrument']=='NSE Index'].copy()
```

```python
df_nsei = df_nsei.sort_values('date')
```

```python
df_nsei = df_nsei.set_index('date')
```

```python
monthly_nsei = df_nsei['close'].resample('M').mean()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))
plt.plot(monthly_nsei)
plt.title('NSE Index – Monthly Average Close Price')
plt.xlabel('Date')
plt.ylabel('Price')
plt.show()
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing
```

```python
hw_model = ExponentialSmoothing(
    monthly_nsei,
    trend='add',        # upward trend hai
    seasonal=None       # seasonality strong nahi
).fit()
```

```python
hw_forecast = hw_model.forecast(12)
```

```python
plt.figure(figsize=(12,5))
plt.plot(monthly_nsei, label='Actual')
plt.plot(hw_forecast, label='Forecast', linestyle='--')
plt.title('NSE Index Forecast – Holt-Winters')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()
```
